```python
survivors = df[df['Survived'] == 1]

print("\n9. Male Survivors:")
print(survivors[survivors['Sex'] == 'male'][['Name', 'Age', 'Sex']])

print("\nFemale Survivors:")
print(survivors[survivors['Sex'] == 'female'][['Name', 'Age', 'Sex']])

# Count male and female survivors
gender_count = survivors['Sex'].value_counts()

print("\nNumber of Male Survivors:")
print(gender_count.get('male', 0))

print("\nNumber of Female Survivors:")
print(gender_count.get('female', 0))

if gender_count.get('female',0) > gender_count.get('male',0):
    print("\nConclusion: More females survived than males.")
else:
    print("\nConclusion: More males survived than females.")

# Count plot for male and female survivors
plt.figure(figsize=(6,4))
sns.countplot(x='Sex', data=survivors)
plt.title("Male vs Female Survivors")
plt.show()

```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Pclass', data=survivors)
plt.title("Survived People by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()

# Display counts
class_count = survivors['Pclass'].value_counts().sort_index()

print("\nSurvivors in each Passenger Class:")
print(class_count)

if class_count.get(1,0) > class_count.get(2,0) and class_count.get(1,0) > class_count.get(3,0):
    print("\nConclusion: More First-Class passengers survived than Second and Third Class passengers.")
else:
    print("\nConclusion: First-Class passengers did not have the highest number of survivors.")
```
